Introducing Data Analysis

Data analysis is a process of inspecting, cleansing, transformation, and modeling data with the goal of discovering useful information, informing conclusions, and supporting decision-making.


Data analysis - It is a detailed examination of the elements of structure of something

Data analytics - It is a systematic computational analysis of something


Types of Data Analysis

Descriptive Analysis - is the most common and primary forms of data analysis. It is used to find the "what is/has happening/happened?" in business. usually, we take to help of descriptive data analysis to track the Key Performance Indicator (KPIs), sales profit/loss, and so on. Example - Publishing weekly sales report

Diagnostic data analysis helps to find out the "why did something happen?" Once we get the report of what happened from descriptive analysis, diagnostic data analysis helps us understand which factors caused something to happen. Example - analyzing the reasons/factors for fewer sales in the previous year

Prescriptive analysis uses the ourcomes of all the data analysis. It will help find the "what action should be taken?". So, it prescribes the action/s as a solution to counter the specific situation. It needs advance machine learning and real-time AI. Example - Dynamic pricing of taxi fares depends on demand, weather, and social conditions.

```python
#Example 1

import pandas as pd
data = {
    'apples':[3, 2, 0, 1],
    'oranges':[0, 3, 7, 2]
}
purchases = pd.DataFrame(data)
print(purchases)
```

```python
#Example 2 DataFrame Index
import pandas as pd
data = {
    'apples':[3, 2, 0, 1],
    'oranges':[0, 3, 7, 2]
}
purchases = pd.DataFrame(data, index=['June', 'Robert', 'Lily', 'David'])
print(purchases)
```

```python
#Example 3 Locate - locate specific record

print(purchases.loc['June'])
```

```python
#Example 4 Importing DataFrame to csv or excel file

purchases.to_csv('purchases.csv') # or purchases.to_excel('purchases.xlsx')
```
