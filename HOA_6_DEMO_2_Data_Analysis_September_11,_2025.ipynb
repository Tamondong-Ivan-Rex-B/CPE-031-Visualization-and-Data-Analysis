Introducing Data Analysis

Data analysis is a process of inspecting, cleansing, transformation, and modeling data with the goal of discovering useful information, informing conclusions, and supporting decision-making.


Data analysis - It is a detailed examination of the elements of structure of something

Data analytics - It is a systematic computational analysis of something


Types of Data Analysis

Descriptive Analysis - is the most common and primary forms of data analysis. It is used to find the "what is/has happening/happened?" in business. usually, we take to help of descriptive data analysis to track the Key Performance Indicator (KPIs), sales profit/loss, and so on. Example - Publishing weekly sales report

Diagnostic data analysis helps to find out the "why did something happen?" Once we get the report of what happened from descriptive analysis, diagnostic data analysis helps us understand which factors caused something to happen. Example - analyzing the reasons/factors for fewer sales in the previous year

Prescriptive analysis uses the ourcomes of all the data analysis. It will help find the "what action should be taken?". So, it prescribes the action/s as a solution to counter the specific situation. It needs advance machine learning and real-time AI. Example - Dynamic pricing of taxi fares depends on demand, weather, and social conditions.

```python
# Interpret your data, appropriate title, appropriate chart

# Display the dataframe of those items with the
#same date of selling,
df.loc[df['OrderDate']== '7/4/22']
```

```python
# Bar chart
date = df[df['OrderDate'] == '7/4/22']

total_units = date['Units'].sum()

# Create a bar chart
plt.figure(figsize=(4, 6))
plt.bar(['7/4/22'], [total_units], color='skyblue', edgecolor='black')

# Add labels and title
plt.title('Total Units Sold on 7/4/22')
plt.xlabel('Order Date')
plt.ylabel('Total Units Sold')

# Show the chart
plt.show()
```

<h1>Descriptive Analysis</h1>
I used a bar graph for this data because the bar chart because this can accurately tell us the total number of items sold for that current day. Also, the chart is expandable in the x-axis, which means that if we want, we can add more dates in order to compare the total number of items sold for different days. Based on the chart provided, we can see that on 7/4/22, Jones had a total item sold of 62 units.


```python
#per region,
df.loc[df['Region']=='Central']
```

```python
#Pie Chart

# Filter data for the Central Region
central_region_df = df[df['Region'] == 'Central']

# Count values in 'Item' for the Central Region and sum the units for each item
item_units = central_region_df.groupby('Item')['Units'].sum()

# Create a pie chart
item_units.plot.pie(figsize=(8, 8), autopct='%1.1f%%')

# Add title
plt.title('Total Units Sold per Item in Central Region')

# Remove y-axis label
plt.ylabel('')

# Show chart
plt.show()
```

<h1>Descriptive Analysis</h1>

For this type of data, I used a pie chart because I noticed that the the total number of units is composed of one category, which is a region. So, like pie, we can split the units from a whole pie to get our percentages of slice or of data. The whole pie will be the whole central region, and the split or distribution of the slices are based on the number of units sold. Based from the pie graph provided, we can conclude that

```python
#Line Chart

# Filter only 'Name of the Representative'
rep = df[df['Rep'] == 'Morgan']

# Group by Orderdate and sum Total
date_total = rep.groupby('OrderDate')['Total'].sum()

# Plot line chart
plt.figure(figsize=(12, 6))
plt.plot(date_total.index, date_total.values, marker='o', linestyle='-', color='blue')

# Add labels and title
plt.title('Total Sales of Morgan in the Central Region')
plt.xlabel('Date of Selling')
plt.ylabel('Total Sales in a Day')

# Rotate x-axis labels for readability
plt.xticks(rotation=90)

# Show chart
plt.show()



```

<h1>Descriptive Analysis</h1>

  In this chart, we can see that as time goes by, Morgan is earning more money. With this data, we can se that they are having higher KPIs, and they are performing better and better in selling items. Another interpretation is that Morgan obtains more connnections as time goes by and takes advantage of that by selling items to those connections. I used a line chart for this example because it can clearly show us the trend of the sales of Morgan.
