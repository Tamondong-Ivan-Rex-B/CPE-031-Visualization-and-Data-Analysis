**4. Procedures:**

Step 1. Import Libraries and Create Sample Data

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Sample sales dataset
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [200, 250, 180, 300, 280, 350],
    "Marketing_Spend": [50, 60, 45, 70, 65, 80]
}
df = pd.DataFrame(data)
df
```

Step 2. Know Your Audience — Compare Chart Styles

```python
# Executive audience: prefer clarity (simple bar)
plt.figure(figsize=(7,4))
sns.barplot(x="Month", y="Sales", data=df, color="steelblue")
plt.title("Executive View: Monthly Sales Overview", fontsize=13)
plt.show()

# Analyst audience: prefer detailed trends (line chart)
plt.figure(figsize=(7,4))
sns.lineplot(x="Month", y="Sales", data=df, marker="o", label="Sales")
sns.lineplot(x="Month", y="Marketing_Spend", data=df, marker="o", label="Marketing Spend")
plt.title("Analyst View: Sales vs Marketing Trend", fontsize=13)
plt.legend()
plt.show()
```

**Task 1:** Create one additional visualization suited for a different audience (e.g., a customer, investor, or teammate).

```python
#Code Here

# Executive audience: prefer clarity (simple bar)
plt.figure(figsize=(7,4))
sns.barplot(x="Month", y="Sales", data=df, color="steelblue")
plt.title("Executive View: Monthly Sales Overview", fontsize=13)
plt.show()

# Analyst audience: prefer detailed trends (line chart)
plt.figure(figsize=(7,4))
sns.lineplot(x="Month", y="Sales", data=df, marker="o", label="Sales")
sns.lineplot(x="Month", y="Marketing_Spend", data=df, marker="o", label="Marketing Spend")
plt.title("Analyst View: Sales vs Marketing Trend", fontsize=13)
plt.legend()
plt.show()

# Investor audience: prefer growth trends (line chart with focus on overall trend)
plt.figure(figsize=(7,4))
sns.lineplot(x="Month", y="Sales", data=df, marker="o", color="green")
plt.title("Investor View: Sales Growth Trend", fontsize=13)
plt.ylabel("Sales ($)")
plt.show()
```

Step 3. Know Your Message — Highlight What Matters

```python
# Highlight the highest sales month (preattentive color cue)
highlight = df["Sales"].idxmax()

colors = ["gray" if i != highlight else "crimson" for i in range(len(df))]
plt.figure(figsize=(7,4))
plt.bar(df["Month"], df["Sales"], color=colors)
plt.title("Highlighting the Best Month in Sales", fontsize=13)
plt.show()
```

**Task 2:** Use color or annotations to emphasize a key insight (e.g., lowest point, sudden increase).

```python
#Code Here

# Highlight the sudden increase in sales in April
plt.figure(figsize=(7,4))
sns.lineplot(x="Month", y="Sales", data=df, marker="o", color="blue")
plt.title("Highlighting the Sales Increase in April", fontsize=13)

# Add annotation for the April increase
plt.annotate("Sudden Increase", xy=("Apr", df.loc[df["Month"] == "Apr", "Sales"].iloc[0]),
             xytext=("Mar", df.loc[df["Month"] == "Mar", "Sales"].iloc[0] + 50),
             arrowprops=dict(facecolor='black', shrink=0.05))

plt.show()
```

Step 4. Storytelling with Charts — Beginning, Middle, End

```python
# Beginning: Context
plt.figure(figsize=(7,4))
sns.barplot(x="Month", y="Sales", data=df, color="skyblue")
plt.title("Beginning: Sales Started Steady", fontsize=13)
plt.show()

# Middle: Change or Challenge
plt.figure(figsize=(7,4))
sns.lineplot(x="Month", y="Sales", data=df, color="orange", marker="o")
plt.title("Middle: A Drop in March Due to Low Marketing", fontsize=13)
plt.show()

# End: Resolution
plt.figure(figsize=(7,4))
sns.barplot(x="Month", y="Sales", data=df, color="limegreen")
plt.title("End: Recovery and Growth After March", fontsize=13)
plt.show()
```

**Task 3:** Write one or two lines describing your data story from beginning → middle → end.

*Answer here <br>
Beginning: Sales started steady in the early months. <br>
Middle: There was a dip in sales in March, possibly linked to lower marketing spend. <br>
End: Sales recovered and showed growth after March, indicating a positive response to increased marketing efforts

Step 5. Bringing It All Together — Presentation Design

```python
# Simple combined story chart
plt.figure(figsize=(8,5))
sns.lineplot(x="Month", y="Sales", data=df, marker="o", linewidth=2, color="royalblue", label="Sales")
sns.lineplot(x="Month", y="Marketing_Spend", data=df, marker="o", linestyle="--", color="orange", label="Marketing Spend")

plt.fill_between(df["Month"], df["Sales"], alpha=0.1, color="royalblue")
plt.title("From Effort to Impact: Marketing Spend Drives Sales Growth", fontsize=13)
plt.legend()
plt.show()
```

```python
#Supplementary Activity
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

file_path = '/content/drive/Othercomputers/My Laptop/TIP QC/TIP/Computer Engineering/2ND YEAR/1ST SEMESTER/CPE 031 Visualizations and Data Analysis/Finals/exams.csv'

df_exams = pd.read_csv(file_path)

# Let's visualize the average math score by parental level of education
avg_math_score = df_exams.groupby('parental level of education')['math score'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 6))

# Create the bar chart using matplotlib
bars = plt.bar(avg_math_score.index, avg_math_score.values, color='skyblue')

# Highlight the highest average math score
highest_score_level = avg_math_score.idxmax()
bars[avg_math_score.index.get_loc(highest_score_level)].set_color('green')

# Highlight the lowest average math score
lowest_score_level = avg_math_score.idxmin()
bars[avg_math_score.index.get_loc(lowest_score_level)].set_color('red')

# Add title and annotation
plt.title('Average Math Score by Parental Level of Education', fontsize=16)
plt.ylabel('Average Math Score')
plt.xlabel('Parental Level of Education')
plt.xticks(rotation=45, ha='right')

# Add annotation for the highest value
highest_score_value = avg_math_score.max()
plt.annotate(f'{highest_score_value:.2f}',
             xy=(avg_math_score.index.get_loc(highest_score_level), highest_score_value),
             xytext=(avg_math_score.index.get_loc(highest_score_level), highest_score_value + 2),
             ha='center',
             fontsize=10,
             color='black')

# Add annotation for the lowest value
lowest_score_value = avg_math_score.min()
plt.annotate(f'{lowest_score_value:.2f}',
             xy=(avg_math_score.index.get_loc(lowest_score_level), lowest_score_value),
             xytext=(avg_math_score.index.get_loc(lowest_score_level), lowest_score_value + 2),
             ha='center',
             fontsize=10,
             color='black')


plt.tight_layout()
plt.show()
```
